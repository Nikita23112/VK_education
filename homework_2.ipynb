## Слияние двух отсортированных списков

```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def mergeTwoLists(list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
    # Создаем фиктивный узел (dummy), чтобы не обрабатывать отдельно голову списка
    dummy = ListNode()
    tail = dummy
    
    # Проходим по обоим спискам, пока они не закончатся
    while list1 and list2:
        if list1.val <= list2.val:
            # Присоединяем узел из list1
            tail.next = list1
            list1 = list1.next
        else:
            # Присоединяем узел из list2
            tail.next = list2
            list2 = list2.next
        # Двигаем хвост вперед
        tail = tail.next

    # Если один из списков закончился, присоединяем остаток другого.
    # Так как списки отсортированы, остаток уже упорядочен.
    tail.next = list1 if list1 else list2
    
    # Возвращаем реальную голову списка (следующий узел после dummy)
    return dummy.next
```
