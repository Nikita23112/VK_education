## 1. Развернуть односвязный список 

```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def reverseList(head: Optional[ListNode]) -> Optional[ListNode]:
    # ПОЧЕМУ АЛГОРИТМ РАБОТАЕТ ЗА O(N):
    # 1. Временная сложность O(N): мы проходим по односвязному списку ровно один раз. 
    #    В цикле для каждого из N узлов выполняется фиксированное, константное число 
    #    операций по перенаправлению указателей (O(1) на один узел). Общее время прямо 
    #    пропорционально количеству элементов N.
    # 2. Пространственная сложность O(1): мы не используем дополнительные структуры данных 
    #    (массивы, списки или рекурсивный стек), размер которых зависит от длины списка. 
    #    В памяти хранится только фиксированное число указателей (prev, cur, nex), 
    #    поэтому затраты памяти постоянны независимо от размера N.
    
    # Случай 1: Список пуст
    if head is None:
        return None
        
    # Случай 2: В списке всего один элемент
    if head.next is None:
        return head
        
    # Случай 3: В списке ровно два элемента (оптимизация / ручной разворот)
    if head.next.next is None:
        temp = head.next       # Сохраняем второй элемент
        head.next = None       # Первый теперь указывает на None (становится хвостом)
        temp.next = head       # Второй указывает на первый (становится новой головой)
        return temp

    # Общий случай: в списке три и более элементов
    prev = None  # Указатель на предыдущий узел (изначально None)
    cur = head   # Указатель на текущий узел (начинаем с головы)
    nex = cur.next # Указатель на следующий узел (чтобы не потерять остаток списка)

    # Проходим по списку, пока следующий узел существует
    while not nex is None:
        cur.next = prev    # Разворачиваем указатель текущего узла назад
        prev = cur         # Сдвигаем prev на текущий узел
        cur = nex          # Сдвигаем cur на следующий узел
        nex = nex.next     # Сдвигаем nex ещё на один шаг вперёд

    # Обрабатываем последний узел списка
    cur.next = prev        
    return cur             # Возвращаем новую голову развернутого списка

# алгоритм работает за O(n) 
```

## 2. Найти середину списка 

```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def findMiddle(head: Optional[ListNode]) -> Optional[ListNode]:
    # Если список пуст, возвращаем None
    if head is None:
        return None

    # Метод двух указателей («черепаха и заяц»):
    # - slow ("медленный") делает 1 шаг за итерацию
    # - fast ("быстрый") делает 2 шага за итерацию
    #
    # ПОЧЕМУ ЭТО РАБОТАЕТ ЗА ОДИН ПРОХОД (O(N)):
    # 1. Соотношение скоростей: fast движется ровно в 2 раза быстрее, чем slow. 
    #    В любой момент времени пройденное fast расстояние вдвое больше расстояния slow.
    # 2. Момент финиша: когда fast добегает до конца списка (проходит всю дистанцию N), 
    #    slow успевает преодолеть ровно половину пути (N / 2).
    # 3. Это избавляет нас от необходимости предварительно вычислять длину списка 
    #    в отдельном цикле, позволяя найти середину «на лету».
    
    slow = head
    fast = head

    # Продолжаем цикл, пока быстрый указатель и его следующий элемент существуют
    while fast is not None and fast.next is not None:
        slow = slow.next          # Медленный сдвигается на 1 шаг
        fast = fast.next.next     # Быстрый сдвигается на 2 шага

    # Когда fast достигает конца (при нечетной длине) или выходит за границы (при четной),
    # slow оказывается ровно посередине списка и возвращается в качестве результата.
    return slow
```

## 3. Удалить элемент из односвязного списка 

```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def removeElements(head: Optional[ListNode], val: int) -> Optional[ListNode]:
    # Создаем фиктивный узел (dummy node), который указывает на голову нашего списка.
    # Это позволяет унифицировать обработку: удаление головы теперь ничем не отличается 
    # от удаления внутреннего элемента, так как у любой реальной головы появляется «предок» (dummy).
    dummy = ListNode(0, head)
    
    # Указатель current начинает с фиктивного узла
    current = dummy

    # Проходим по списку, проверяя следующий элемент
    while current.next is not None:
        if current.next.val == val:
            # Если значение следующего узла равно val, мы «перешагиваем» его,
            # перенаправляя ссылку в обход (удаление in-place)
            current.next = current.next.next
        else:
            # Иначе просто продвигаемся дальше по списку
            current = current.next

    # Возвращаем реальную новую голову списка, которая теперь хранится в dummy.next
    return dummy.next
```

## 4. Является ли одна строка исходной для другой

```python
def isOriginalString(a: str, b: str) -> bool:
    # Если исходная строка длиннее результирующей, она явно не могла быть в неё превращена добавлением символов
    if len(a) > len(b):
        return False
        
    # Инициализируем два указателя:
    # i — для строки a (исходная)
    # j — для строки b (полученная добавлением символов)
    i = 0
    j = 0
    
    # Проходим по обеим строкам с помощью указателей
    while i < len(a) and j < len(b):
        # Если символы совпадают, значит, текущий символ из исходной строки a
        # нашелся в строке b. Передвигаем указатель i на следующий символ строки a.
        if a[i] == b[j]:
            i += 1
        
        # Указатель j результирующей строки b сдвигается на каждом шаге в любом случае,
        # перебирая все добавленные и исходные символы.
        j += 1
        
    # Если указатель i дошел до конца строки a, значит, все символы строки a 
    # были успешно найдены в строке b в правильном порядке.
    return i == len(a)
```

## 5. Метод двух указателей (поиск пары) 

```python
from typing import List

def twoSumSorted(arr: List[int], target: int) -> bool:
    # Инициализируем два указателя:
    # left — указывает на самый первый (минимальный) элемент массива
    # right — указывает на самый последний (максимальный) элемент массива
    left = 0
    right = len(arr) - 1

    # ПОЧЕМУ АЛГОРИТМ РАБОТАЕТ ЗА O(N):
    # 1. Один проход навстречу друг другу: на каждой итерации либо левый указатель 
    #    сдвигается вправо (left += 1), либо правый влево (right -= 1). За каждый шаг 
    #    расстояние между ними сокращается как минимум на 1. В худшем случае они 
    #    суммарно совершат не более N шагов, пройдя весь массив.
    # 2. Константное время на один шаг O(1): внутри цикла выполняются только простые 
    #    арифметические операции и сравнения, не зависящие от размера массива.
    # 3. Отсутствие вложенных циклов: благодаря тому, что массив уже отсортирован, 
    #    мы избегаем полного перебора всех пар (который давал бы O(n^2)), а точно 
    #    направляем указатели к цели.
    # 
    # Пространственная сложность O(1): мы не создаем дополнительные структуры данных 
    # и используем только две переменные для указателей.
    while left < right:
        current_sum = arr[left] + arr[right]

        if current_sum == target:
            # Нашли пару элементов, сумма которых равна target
            return True
        elif current_sum < target:
            # Если сумма меньше целевой, сдвигаем левый указатель вправо (к большим числам)
            left += 1
        else:
            # Если сумма больше целевой, сдвигаем правый указатель влево (к меньшим числам)
            right -= 1

    # Если указатели сошлись, а пара не нашлась, возвращаем False
    return False
```

## 6. Является ли слово палиндромом? 

```python
def isPalindrome(s: str) -> bool:
    # Инициализируем два указателя:
    # left — указывает на самый первый символ строки (начало)
    # right — указывает на самый последний символ строки (конец)
    left = 0
    right = len(s) - 1

    # КАК РАБОТАЕТ СЛОЖНОСТЬ И СРАВНЕНИЕ:
    # 1. Временная сложность O(n): Указатели движутся навстречу друг другу от краев к центру. 
    #    Каждый символ строки участвует в сравнении не более одного раза. 
    #    Суммарное число шагов пропорционально длине строки n, а внутри цикла 
    #    выполняются базовые операции за константное время O(1).
    # 2. Пространственная сложность O(1): Мы не создаем копии строк или новые структуры данных, 
    #    а используем только два указателя, затрачивая минимум памяти.
    # 3. Сравнение символов: На каждом шаге мы сравниваем символы s[left] и s[right]. 
    #    Если они не совпадают, строка точно не является палиндромом (возвращаем False). 
    #    Если совпадают, мы сдвигаем оба указателя ближе к центру (left += 1, right -= 1), 
    #    проверяя следующие симметричные позиции.
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1

    # Если указатели встретились или пересеклись, а несоответствий не нашлось — это палиндром
    return True
```

## 7. Метод двух указателей (удаление дубликатов) 

```python
from typing import List

def removeDuplicates(nums: List[int]) -> int:
    # Если массив пуст, уникальных элементов нет
    if not nums:
        return 0

    # Инициализируем два указателя:
    # slow ("медленный") — указывает на позицию последнего уникального элемента.
    # fast ("быстрый") — сканирует весь массив в поисках новых уникальных элементов.
    slow = 0

    # СЛОЖНОСТЬ И ПРИНЦИП РАБОТЫ IN-PLACE:
    # 1. Временная сложность O(n): Быстрый указатель проходит по всему массиву ровно один раз 
    #    (цикл от 1 до len(nums)). Все проверки и операции выполняются за константное время O(1).
    # 2. Пространственная сложность O(1): Никаких дополнительных массивов или структур данных 
    #    не создается. Изменения происходят строго внутри исходного массива (in-place).
    # 3. Как работает перезапись без дополнительной памяти: 
    #    Когда fast находит элемент, отличный от текущего уникального (nums[fast] != nums[slow]), 
    #    мы сдвигаем slow на один шаг вперед (slow += 1) и перезаписываем значение в ячейке slow 
    #    новым уникальным значением из fast (nums[slow] = nums[fast]). Таким образом, все уникальные 
    #    элементы постепенно «спрессовываются» в начале массива, а дубликаты перезаписываются «на лету».
    for fast in range(1, len(nums)):
        # Если нашли новый уникальный элемент
        if nums[fast] != nums[slow]:
            slow += 1              # Продвигаем границу уникальных элементов
            nums[slow] = nums[fast]  # Перезаписываем дубликат на актуальное уникальное значение

    # Возвращаем количество уникальных элементов (индекс slow + 1)
    return slow + 1
```

## 8. Слияние двух отсортированных списков 

```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def mergeTwoLists(list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
    # Создаем фиктивный узел (dummy), чтобы не обрабатывать отдельно голову списка
    dummy = ListNode()
    tail = dummy
    
    # СЛОЖНОСТЬ И АНАЛИЗ ПАМЯТИ:
    # 1. Временная сложность O(n + m): Где n и m — длины первого и второго списков. 
    #    В цикле на каждой итерации мы берем ровно один узел из list1 или list2, 
    #    поэтому общее число шагов равно сумме элементов обоих списков (n + m). 
    #    Все операции внутри цикла выполняются за константное время O(1).
    # 2. Пространственная сложность O(1) (без дополнительной памяти): 
    #    Мы не создаем новые узлы ListNode (кроме одного временного фиктивного узла dummy, 
    #    который не зависит от размеров списков). Слияние происходит строго *in-place* 
    #    за счет простого перенаправления существующих ссылок (указателей next).
    
    # Проходим по обоим спискам, пока они не закончатся
    while list1 and list2:
        if list1.val <= list2.val:
            # Присоединяем узел из list1
            tail.next = list1
            list1 = list1.next
        else:
            # Присоединяем узел из list2
            tail.next = list2
            list2 = list2.next
        # Двигаем хвост вперед
        tail = tail.next

    # Если один из списков закончился, присоединяем остаток другого.
    # Так как списки отсортированы, остаток уже упорядочен.
    tail.next = list1 if list1 else list2
    
    # Возвращаем реальную голову списка (следующий узел после dummy)
    return dummy.next
```
