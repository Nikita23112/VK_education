## Two sum

```python
from typing import List

def twoSum(numbers: List[int], target: int) -> List[int]:
    l = 0
    r = len(numbers) - 1
    while l < r:
        if numbers[l] + numbers[r] == target:
            return [l, r]
        elif numbers[l] + numbers[r] > target:
            r -= 1
        else:
            l += 1
    return []
```

## Развернуть массив

```python
from typing import List

def reverseString(arr: List[int]) -> List[int]:
    l = 0
    r = len(arr) - 1
    while l < r:
        arr[l], arr[r] = arr[r], arr[l]
        l += 1
        r -= 1

    return arr
```

## Развернуть часть массива


```python
from typing import List

def rotate(nums: List[int], k: int) -> List[int]:
    k = k%len(nums)
    i = 0
    j = len(nums) - 1
    while i < j:
        nums[i], nums[j] = nums[j], nums[i]
        i += 1
        j -= 1
    i = 0
    j = k - 1
    while i < j:
        nums[i], nums[j] = nums[j], nums[i]
        i += 1
        j -= 1
    i = k
    j = len(nums) - 1
    while i < j:
        nums[i], nums[j] = nums[j], nums[i]
        i += 1
        j -= 1

    return nums
```

## Слияние двух отсортированных массивов

```python
from typing import List

def merge(nums1: List[int], nums2: List[int]) -> List[int]:
    m = len(nums1)
    n = len(nums2)
    i = m-1
    j = n-1
    k = m + n - 1
    while i >= 0 and j >= 0:
        if nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1
        k -= 1
    while j >= 0:
        nums1[k] = nums2[j]
        j -= 1
        k -= 1

    return nums1
```

## Минимальный размер подмассива

```python
from typing import List

def minSubArrayLen(target: int, nums: List[int]) -> int:
    l, total = 0, 0
    res = float('inf')

    for r in range(len(nums)):
        total += nums[r]
        while total >= target:
            res = min(res, r-l+1)
            total -= nums[l]
            l += 1
            
    return 0 if res == float('inf') else res
```

## Сортировка массива из 0 и 1

```python
from typing import List

def sort_binary_array(arr: List[int]) -> List[int]:
    left = 0
    right = len(arr) - 1
    i = 0
    
    while i <= right:
        if arr[i] == 0:
            arr[i], arr[left] = arr[left], arr[i]
            left += 1
            i += 1
        elif arr[i] == 1:
            arr[i], arr[right] = arr[right], arr[i]
            right -= 1
        else:
            i += 1
    
    return arr
```

## Задача флага Нидерландов

```python
from typing import List

def sort_colors(arr: List[int]) -> List[int]:
    low = 0
    mid = 0
    high = len(arr) - 1
    
    while mid <= high:
        if arr[mid] == 0:
            arr[low], arr[mid] = arr[mid], arr[low]
            low += 1
            mid += 1
        elif arr[mid] == 1:
            mid += 1
        else: 
            arr[mid], arr[high] = arr[high], arr[mid]
            high -= 1
    
    return arr
```

## Передвинуть четные числа вперед


```python
from typing import List

def even_first(arr: List[int]) -> List[int]:
    even_index = 0
    for i in range(len(arr)):
        if arr[i] % 2 == 0:
            arr[i], arr[even_index] = arr[even_index], arr[i]
            even_index += 1
    return arr
```

## Нули в конец массива

```python
def move_zeros_to_end(arr: List[int]) -> List[int]:
    write = 0  # куда положим следующий ненулевой
    for i in range(len(arr)):
        if arr[i] != 0:
            arr[i], arr[write] = arr[write], arr[i]
            write += 1
    return arr
```
